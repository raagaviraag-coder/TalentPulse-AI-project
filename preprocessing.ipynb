```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load dataset
df = pd.read_csv("ml_project_dataset.csv")

print(df.shape)
df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
df.drop(
    ["EmployeeCount", "EmployeeNumber", "Over18", "StandardHours"],
    axis=1,
    inplace=True
)

print(df.shape)
df.head()
```

```python
le = LabelEncoder()

df["Attrition"] = le.fit_transform(df["Attrition"])

df.head()
```

```python
df = pd.get_dummies(df, drop_first=True)

print("Dataset Shape:", df.shape)
df.head()
```

```python
print(df.dtypes)
```

```python
df.to_csv("ml_project_dataset_cleaned.csv", index=False)

print("✅ Cleaned dataset saved successfully!")
```
